### Data Preprocessing: Adding New Columns and Handling Duplicates

```python

df['SepalArea'] = df['SepalLengthCm'] * df['SepalWidthCm']

print('--- Iris DataFrame with new SepalArea column (first 5 rows) ---')
display(df.head())


df['PetalArea'] = df['PetalLengthCm'] * df['PetalWidthCm']

print('\n--- Iris DataFrame with new PetalArea column (first 5 rows) ---')
display(df.head())
```

```python

duplicates_df = df.duplicated()
print('--- Number of duplicate rows in Iris dataset ---')
print(duplicates_df.sum())


duplicates_sample_df = sample_df.duplicated()
print('\n--- Number of duplicate rows in Sample DataFrame ---')
print(duplicates_sample_df.sum())
```

```python

df_renamed = df.rename(columns={'SepalLengthCm': 'Sepal_Length_Cm', 'PetalLengthCm': 'Petal_Length_Cm'})

print('--- Iris DataFrame with renamed columns (first 5 rows) ---')
display(df_renamed.head())
```
